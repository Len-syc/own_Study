```python
from typing import Any


class Shopping:
    def __init__(self,items:list[str]):
        self.items = items
    
    def __add__(self,other_Car):
        New_Car = Shopping(self.items + other_Car.items)
        return New_Car
    def __str__(self):
        return f"{self.items}"
    def __len__(self):
        return len(self.items)
    def __call__(self,items):
        self.items.append(items)
c1 = Shopping(["1","2"])
c2 = Shopping(["3","4"])
c3 = c1+c2
c3("5")
print(c3)
# print(c2)
print(len(c3))
```

```python
class add:
    def __init__(self,numbers):
        self.numbers = numbers
    def __call__(self,number):
        return add(self.numbers+number)
    def __add__(self,item):
        return self.numbers + item
    def __str__(self):
        return f"{self.numbers}"
    
addTwo = add(2)
print(addTwo)#2
print(addTwo + 5) #7
print(addTwo(3)) #5
print(addTwo(3)(5)) #10
```
