```python
#面向对象 类方法
class Students:
    Students_num = 0
    def __init__(self,name,sex):
        self.name = name
        self.sex  = sex
        self.Students_num+=1
    def show(self):
        print(f"self.name:{self.name},self.sex:{self.sex}")
        print(self.Students_num)
    @classmethod
    def Add_Students(cls,num):
        cls.Students_num += num
    @classmethod
    def from_String(cls,info):
        name,sex = info.split(",")
        return cls(name,sex)
a1 = Students.from_String("Lenqin,male")
a1.show()
```

```python
#静态方法
class Students:
    Students_num = 0
    def __init__(self,name,sex):
        self.name = name
        self.sex  = sex
        self.Students_num+=1
    def show(self):
        print(f"self.name:{self.name},self.sex:{self.sex}")
        print(self.Students_num)
    @classmethod
    def Add_Students(cls,num):
        cls.Students_num += num
    @classmethod
    def from_String(cls,info):
        name,sex = info.split(",")
        return cls(name,sex)
    @staticmethod
    def Re_Len(name): #调用的时候要用类,Students.Re_Len()
        if isinstance(name,str):
            return len(name)
        return -1
a1 = Students.from_String("Lenqin,male")
a1.show()
print(f"len:{Students.Re_Len(a1.name)}")
```
